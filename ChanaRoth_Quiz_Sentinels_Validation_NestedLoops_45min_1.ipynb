## Problem 2 — Character Analyzer (≈12 minutes)

Ask the user to enter **one word**.

Use a `for` loop to examine **each character** in the word.

For every character, use an inner decision (`if` / `elif` / `else`) to determine whether it is:

- a vowel: `a, e, i, o, u`
- a digit: `0` through `9`
- something else

Keep three counters and print all three totals at the end.

Hint : use the `in` operator Usage :
`
guess in word
`

### Example
If the user enters:

```text
Room2
```

the program should report:

```text
Vowels: 2
Digits: 1
Other characters: 2
```

### Important
Uppercase and lowercase vowels should both count as vowels.

**Write your complete solution in the code cell below.**

````python
# PROBLEM 2 — Write your code below
word = input("Enter one word: ")
vowel = 0
digit = 0
other = 0
for ch in word:
  if ch.lower() in {'a', 'e', 'i', 'o','u'}:
      vowel = vowel + 1
  elif ch in {'0','1','3','4','5','6','7','8','9'}:
    digit = digit + 1
  else:
    other = other + 1

print(f"vowels: {vowel}")
print(f"Digits: {digit}")
print(f"Other characters: {other}")
````
